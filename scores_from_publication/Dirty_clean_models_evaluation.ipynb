```python
from pathlib import Path
import cv2
import numpy as np
from ultralytics import YOLO

# ===== KONFIGURACJA =====
SOURCE_ROOT = Path(r"D:\Politechnika\Deadlift_publikacja_nr_2\spinal_pose_walidacja\spinal_pose_1000_split")
MODEL_BRUDNY_PATH = Path(r"D:\Politechnika\Deadlift_publikacja_nr_2 — kopia\Skrypty\github\weryfikacje\notatniki_weryfikacyjne_sprawiedliwe\brudny_model_treningowy.pt")
MODEL_CZYSTY_PATH = Path(r"D:\Politechnika\Deadlift_publikacja_nr_2 — kopia\Skrypty\github\weryfikacje\notatniki_weryfikacyjne_sprawiedliwe\czysty_model_treningowy_2.pt")
IMAGE_EXTS = {".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff", ".webp"}
PRED_CONF_THR = 0.20
EXPECTED_NUM_KP = 9

KEYPOINT_NAMES = [
    "kp_00", "kp_01", "kp_02", "kp_03", "kp_04",
    "kp_05", "kp_06", "kp_07", "kp_08",
]

images_root = SOURCE_ROOT / "images"
labels_root = SOURCE_ROOT / "labels"

def parse_gt_keypoints_from_label(label_path):
    """Zwraca keypointy GT jako tablice (N, 2) w skali [0,1]."""
    if not label_path.exists():
        return None

    lines = [ln.strip() for ln in label_path.read_text(encoding="utf-8").splitlines() if ln.strip()]
    if not lines:
        return None

    vals = [float(v) for v in lines[0].split()]
    if len(vals) < 8:
        return None

    rest = vals[5:]
    if len(rest) % 3 != 0:
        return None

    pts = []
    for i in range(0, len(rest), 3):
        x, y, vis = rest[i : i + 3]
        pts.append((x, y) if vis > 0 else (np.nan, np.nan))

    arr = np.array(pts, dtype=np.float32)
    return arr if arr.size > 0 else None

def predict_keypoints_norm(model, img, conf_thr=0.20):
    """Zwraca keypointy predykcji jako tablice (N, 2) w skali [0,1]."""
    pred = model.predict(source=img, conf=conf_thr, verbose=False)[0]
    kpts = pred.keypoints
    if kpts is None or kpts.xy is None or len(kpts.xy) == 0:
        return None

    h, w = img.shape[:2]
    best_idx = 0
    if pred.boxes is not None and pred.boxes.conf is not None and len(pred.boxes.conf) > 0:
        best_idx = int(np.argmax(pred.boxes.conf.cpu().numpy()))

    xy = kpts.xy.cpu().numpy()[best_idx]
    xy_norm = np.zeros_like(xy, dtype=np.float32)
    xy_norm[:, 0] = xy[:, 0] / max(1, w)
    xy_norm[:, 1] = xy[:, 1] / max(1, h)
    return xy_norm

def accumulate_errors(gt_xy, pred_xy, sums, counts):
    if gt_xy is None or pred_xy is None:
        return

    n = min(len(gt_xy), len(pred_xy), len(sums), EXPECTED_NUM_KP)
    if n == 0:
        return

    gt = gt_xy[:n]
    pred = pred_xy[:n]
    valid = (~np.isnan(gt[:, 0])) & (~np.isnan(gt[:, 1])) & (~np.isnan(pred[:, 0])) & (~np.isnan(pred[:, 1]))

    if not np.any(valid):
        return

    d = np.sqrt((gt[valid, 0] - pred[valid, 0]) ** 2 + (gt[valid, 1] - pred[valid, 1]) ** 2)
    valid_idx = np.where(valid)[0]
    for idx_local, err in zip(valid_idx, d):
        sums[idx_local] += float(err)
        counts[idx_local] += 1

print("Ladowanie modeli...")
model_brudny = YOLO(str(MODEL_BRUDNY_PATH))
model_czysty = YOLO(str(MODEL_CZYSTY_PATH))

image_paths = sorted([p for p in images_root.rglob("*") if p.is_file() and p.suffix.lower() in IMAGE_EXTS])

# Sprawdzenie liczby keypointow w danych
detected_num_kp = None
for img_path in image_paths:
    rel = img_path.relative_to(images_root)
    label_path = (labels_root / rel).with_suffix(".txt")
    gt = parse_gt_keypoints_from_label(label_path)
    if gt is not None:
        detected_num_kp = len(gt)
        break

if detected_num_kp is None:
    raise RuntimeError("Nie znaleziono poprawnych etykiet keypointow w labels.")

if detected_num_kp != EXPECTED_NUM_KP:
    print(f"UWAGA: wykryto {detected_num_kp} keypointow w etykietach, ale licze pierwsze {EXPECTED_NUM_KP}.")

num_kp = EXPECTED_NUM_KP
if len(KEYPOINT_NAMES) != num_kp:
    KEYPOINT_NAMES = [f"kp_{i:02d}" for i in range(num_kp)]

sum_b = np.zeros(num_kp, dtype=np.float64)
sum_c = np.zeros(num_kp, dtype=np.float64)
cnt_b = np.zeros(num_kp, dtype=np.int64)
cnt_c = np.zeros(num_kp, dtype=np.int64)

compared_images = 0
skipped_missing_label = 0
skipped_read_error = 0

for i, img_path in enumerate(image_paths, start=1):
    rel = img_path.relative_to(images_root)
    label_path = (labels_root / rel).with_suffix(".txt")

    if not label_path.exists():
        skipped_missing_label += 1
        continue

    img = cv2.imread(str(img_path))
    if img is None:
        skipped_read_error += 1
        continue

    gt_xy = parse_gt_keypoints_from_label(label_path)
    if gt_xy is None:
        continue

    pred_b = predict_keypoints_norm(model_brudny, img, conf_thr=PRED_CONF_THR)
    pred_c = predict_keypoints_norm(model_czysty, img, conf_thr=PRED_CONF_THR)

    accumulate_errors(gt_xy, pred_b, sum_b, cnt_b)
    accumulate_errors(gt_xy, pred_c, sum_c, cnt_c)

    compared_images += 1
    if i % 25 == 0:
        print(f"Postep: {i}/{len(image_paths)}")

mean_b = np.divide(sum_b, cnt_b, out=np.full_like(sum_b, np.nan), where=cnt_b > 0)
mean_c = np.divide(sum_c, cnt_c, out=np.full_like(sum_c, np.nan), where=cnt_c > 0)

overall_b = float(np.nanmean(mean_b))
overall_c = float(np.nanmean(mean_c))

print("\n=== PODSUMOWANIE ===")
print(f"Porownane obrazy: {compared_images}")
print(f"Pominiete (brak label): {skipped_missing_label}")
print(f"Pominiete (blad odczytu): {skipped_read_error}")
print(f"Sredni blad globalny - BRUDNY: {overall_b:.6f}")
print(f"Sredni blad globalny - CZYSTY: {overall_c:.6f}")
print("Werdykt globalny: " + ("CZYSTY dokladniejszy" if overall_c < overall_b else "BRUDNY dokladniejszy"))

print("\n=== SREDNI BLAD NA KEYPOINT (9 KP) ===")
print(f"{'keypoint':<12} {'brudny':>12} {'czysty':>12} {'delta(czysty-brudny)':>22} {'n_b':>8} {'n_c':>8}")
print("-" * 80)
for k in range(num_kp):
    delta = mean_c[k] - mean_b[k] if not (np.isnan(mean_c[k]) or np.isnan(mean_b[k])) else np.nan
    mb = "nan" if np.isnan(mean_b[k]) else f"{mean_b[k]:.6f}"
    mc = "nan" if np.isnan(mean_c[k]) else f"{mean_c[k]:.6f}"
    dd = "nan" if np.isnan(delta) else f"{delta:.6f}"
    print(f"{KEYPOINT_NAMES[k]:<12} {mb:>12} {mc:>12} {dd:>22} {cnt_b[k]:>8} {cnt_c[k]:>8}")
```
