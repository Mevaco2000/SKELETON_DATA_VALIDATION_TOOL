```python
def test(x):
    dataset = YOLOPoseDataset(r"D:\Politechnika\Deadlift_publikacja_nr_2\Dane_do_treningu\flat_axis_out_"+str(x)+r"\train.txt")
    validator = YPSetValidation(dataset)
    results_lbp=validator.predict_lbp_anomalies_by_embedding_groups(patch_size=128)
    report_path = os.path.join(dataset_path + f"_axis_out_{x}") + ".txt"
    reference_results_txt_path = os.path.join(dataset_path + f"_axis_out_{x}","results.txt")
    save_distance_anomaly_report(results_lbp, report_path)
    comparison = compare_results_txt_to_distance_anomalies(
        reference_results_txt_path=reference_results_txt_path,
        distance_anomaly_report_path=report_path,
        top_k=x,
        match_mode="auto",
        include_reference_fields=True,
    )
    return comparison
```
