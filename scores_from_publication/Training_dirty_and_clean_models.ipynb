## Porownanie przebiegu treningow na podstawie plikow results.csv

W razie potrzeby zmien sciezki w `RESULTS_FILES`, aby wskazywaly wlasciwe foldery runow.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Mapowanie: czysty = train29, brudny = train28.
RESULTS_FILES = {
    "Czysty trening": "weryfikacje/notatniki_weryfikacyjne_sprawiedliwe/runs/pose/train29/results.csv",
    "Brudny trening": "weryfikacje/notatniki_weryfikacyjne_sprawiedliwe/runs/pose/train28/results.csv",
}


def resolve_project_root() -> Path:
    cwd = Path.cwd()
    if (cwd / "scores_from_publication").exists():
        return cwd
    if (cwd.parent / "scores_from_publication").exists():
        return cwd.parent
    return cwd


project_root = resolve_project_root()

runs = {}
for run_name, rel_path in RESULTS_FILES.items():
    path = project_root / rel_path
    if not path.exists():
        raise FileNotFoundError(f"Nie znaleziono pliku: {path}")
    df = pd.read_csv(path)
    df.columns = [c.strip() for c in df.columns]
    runs[run_name] = df

# IEEE-like styling aligned with previous figures.
plt.rcParams.update({
    "font.family": "serif",
    "font.serif": ["Times New Roman", "Times", "DejaVu Serif"],
    "axes.linewidth": 1.0,
})

color_clean = "#2E8B57"
color_dirty = "#C0392B"
clean_label = "Training on clean samples"
dirty_label = "Training on noisy samples"

clean_df = runs["Czysty trening"]
dirty_df = runs["Brudny trening"]

x_values = clean_df["epoch"].to_numpy(dtype=float)
clean_box = clean_df["train/box_loss"].to_numpy(dtype=float)
dirty_box = dirty_df["train/box_loss"].to_numpy(dtype=float)
clean_pose = clean_df["train/pose_loss"].to_numpy(dtype=float)
dirty_pose = dirty_df["train/pose_loss"].to_numpy(dtype=float)

# One chart with two Y scales: left for box loss, right for pose loss.
fig, ax_box = plt.subplots(1, 1, figsize=(8.5, 8.0))
ax_pose = ax_box.twinx()

line1, = ax_box.plot(
    x_values,
    clean_box,
    marker="o",
    markersize=10,
    linewidth=3.0,
    label=f"{clean_label} - box",
    color=color_clean,
    linestyle="-",
    markerfacecolor="white",
    markeredgewidth=2.0,
    alpha=0.95,
    zorder=3,
)
line2, = ax_box.plot(
    x_values,
    dirty_box,
    marker="s",
    markersize=9,
    linewidth=3.0,
    label=f"{dirty_label} - box",
    color=color_dirty,
    linestyle="--",
    markerfacecolor="white",
    markeredgewidth=1.9,
    alpha=0.95,
    zorder=3,
)
line3, = ax_pose.plot(
    x_values,
    clean_pose,
    marker="^",
    markersize=9,
    linewidth=2.8,
    label=f"{clean_label} - pose",
    color=color_clean,
    linestyle=":",
    markerfacecolor=color_clean,
    markeredgewidth=1.8,
    alpha=0.95,
    zorder=2,
)
line4, = ax_pose.plot(
    x_values,
    dirty_pose,
    marker="D",
    markersize=8,
    linewidth=2.8,
    label=f"{dirty_label} - pose",
    color=color_dirty,
    linestyle="-.",
    markerfacecolor=color_dirty,
    markeredgewidth=1.7,
    alpha=0.95,
    zorder=2,
)

xticks = np.unique(np.linspace(float(np.min(x_values)), float(np.max(x_values)), 6, dtype=int))
ax_box.set_xticks(xticks)
ax_box.set_xticklabels([str(int(v)) for v in xticks], fontsize=19, fontweight="bold")
ax_box.set_xlabel("Epoch", fontsize=22, fontweight="bold")

ax_box.set_ylabel("Box loss", fontsize=24, fontweight="bold")
ax_pose.set_ylabel("Pose loss", fontsize=24, fontweight="bold")

ax_box.tick_params(axis="y", labelsize=19)
ax_pose.tick_params(axis="y", labelsize=19)
for tick in ax_box.get_yticklabels():
    tick.set_fontweight("bold")
for tick in ax_pose.get_yticklabels():
    tick.set_fontweight("bold")

xpad = (float(np.max(x_values)) - float(np.min(x_values))) * 0.05
ax_box.set_xlim(float(np.min(x_values)) - xpad, float(np.max(x_values)) + xpad)

ax_box.set_ylim(0, float(max(np.max(clean_box), np.max(dirty_box))) * 1.15)
ax_pose.set_ylim(0, float(max(np.max(clean_pose), np.max(dirty_pose))) * 1.10)

ax_box.grid(axis="y", linestyle="--", linewidth=0.6, alpha=0.5)
ax_box.set_axisbelow(True)

legend_lines = [line1, line2, line3, line4]
legend_labels = [line.get_label() for line in legend_lines]
ax_box.legend(legend_lines, legend_labels, framealpha=0.95, fontsize=15, loc="upper right")

fig.subplots_adjust(left=0.16, right=0.86, bottom=0.18, top=0.92)
plt.show()

summary_rows = []
for run_name, df in runs.items():
    summary_rows.append(
        {
            "Trening": run_name,
            "Liczba epok": int(df["epoch"].max()),
            "Min train/box_loss": df["train/box_loss"].min(),
            "Min train/pose_loss": df["train/pose_loss"].min(),
        }
    )

pd.DataFrame(summary_rows).round(5)
```
